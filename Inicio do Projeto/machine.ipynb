```python
import numpy as np

# criando um array a partir de uma lista
lista = [1,2,3,4,5,]
array1 = np.array(lista)
print("Array1:", array1)

#cirando um array 2d (matriz)
matriz = np.array([[1,2,3], [4,5,6]])
print("Matriz: \n", matriz)
```

Operações Básicas com Arrays:

```python
#Soma, multiplicações e operações element-wise
array2 = np.array([10,20,30,40,50])
soma = array1 + array2
multiplicacao = array1*array2
print("Soma", soma)
print("Multiplicação", multiplicacao)
```

```python
#Funções Numpy Comuns:
media = np.mean(array2)
desvio = np.std(array2)
print( "Média:", media)
print("Desvio Padrão:", desvio)
```

Importação e Criação de DataFrame:

```python
import pandas as pd

#criando um dataframe a partir de um dicionário
dados = {
    "Nome": ["Alice", "Bob", "Charlie"],
    "Idade": [25, 30, 35],
    "Salário": [5000, 6000, 7000]
}

df = pd.DataFrame(dados)
print("DataFrame:\n", df)
```

Manipulação e Exploração de Dados:

```python
#selecionando uma coluna
print("coluna 'nome':\n", df['Nome'])

#filtrando linhas com base em uma condição
df_filtrado = df[df['Salário'] > 6000]
print("Funcionário com salário > 6000: \n", df_filtrado)

#estatisticas descritivas 
print("Estatisticas Descritivas: \n", df.describe())
```

```python
import matplotlib.pyplot as plt

#Gráfico de linha
plt.plot(df['Idade'], df['Salário'], marker = 'o')
plt.title("Salário x Idade")
plt.xlabel("Idade")
plt.ylabel("Salário")
plt.show()


```

```python
# Criando gráfico de barras

plt.bar(df['Nome'], df['Salário'], color ='skyblue')
plt.title("Salário x Nome")
plt.xlabel("Nome")
plt.ylabel("Salário")
plt.show()
```

```python
# Gáfico de dispersão ( scatter plot)
import seaborn as sns
sns.scatterplot(x='Idade', y="Salário", data=df, hue="Idade", s=150)
plt.title("Scatter plot: Salário vs Idade")
plt.show()
```

```python
sns.boxplot(y='Salário', data=df)
plt.title("Distribuição de Salário")
plt.show()
```

```python
#Gráfico de Histograma com KDE

sns.histplot(df['Salário'], kde=True)
plt.title("Histograma com Salario")
plt.show()
```
