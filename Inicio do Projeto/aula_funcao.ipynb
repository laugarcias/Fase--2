```python
def calcular_media(numeros):
    """
    Calcular a média de uma lista de números.

    Parametros:
    numeros (list): Uma lista de números.

    return:
    float: A média dos números na lista.

    """

    if not numeros:
        return 0.0  # Retorna 0.0 se a lista estiver vazia
    return sum(numeros) / len(numeros)  # Calcula e retorna a média

# uso da funcao
valores= [10,20,30,40,50]
print("Média:", calcular_media(valores))
```
