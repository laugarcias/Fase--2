```python
def categorizar_tomate_media (media):
    if  media >=40 and media <= 70:
        return "tomate médio"
    elif media < 40:
        return "tomate pequeno"
    else:
        return "Tomatão"
    
```

```python
#criando uma coluna nova com base na função anterior 
df_csv["categoria_tomate"] = df_csv["Average"].apply(categorizar_tomate_media)
```

```python
# verificando a coluna criada, ela já aprece nos dados
df_csv.head()
```

```python
df_csv.groupby(["categoria_tomate"]).describe() #criando a estatistica da função criada de tomate
```

```python
# Criando filtros
filtro = df_csv["Average"] < 40
df_csv.loc[filtro]
```

```python
#criando uma matriz
import numpy as np
arr_list = np.array([1.2,3,4,5,6,7,8])
print(arr_list)
```

```python
#criando matriz de zeros
arr_zeros = np.zeros((4,6))
print(arr_zeros)
```

```python
#criando matriz com numeros
arr_ones = np.ones((3,4))
print(arr_ones)
```

```python
#criando matriz com numeros aleatorios
arr_random = np.random.rand(3,4)
print(arr_random)
```

```python
#verificar dimensão do array com shape
print(arr_random.shape)
```

```python
#trocando o formato do array
arr_random_reshape = arr_random.reshape((4,3))
print(arr_random_reshape)
```

```python
# criando varios arrays para juntar
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
arr3 = np.array([[9, 10], [11, 12]])

arr4 = np.concatenate ((arr1, arr2, arr3), axis=1)
arr4
```

```python
#dividindo array
arr4_split = np.split(arr4,2)
print(arr4_split)
```

```python
#transpor array
arr4_transpose = np.transpose(arr4)
print(arr4_transpose)
```

```python
#retirar do transpor
arr4_revertido =arr4_transpose.T
print(arr4_revertido)
```

```python
#soma de array
arr_a = np.array([1, 7, 27])
arr_b = np.array([1, 5, 1])
arr_a_b = np.add(arr_a, arr_b)
print(arr_a_b)
```

```python
#subtração de array
arr_sub_a_b = np.subtract(arr_a, arr_b)
print(arr_sub_a_b)
```
